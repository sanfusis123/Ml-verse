```python
from transformers import AutoTokenizer, AutoModelForCausalLM

# Load the model and tokenizer
model_name = "gpt2"  # Can be replaced with any language model like phi-3
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Define a prompt
prompt = "Once upon a time, in a land far, far away"

# Encode and generate text
inputs = tokenizer(prompt, return_tensors='pt')
outputs = model.generate(inputs.input_ids, max_length=50, num_return_sequences=1)

# Decode the generated text
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print("Generated Text:", generated_text)
```

```python
from transformers import pipeline

# Load the text classification pipeline
classifier = pipeline("sentiment-analysis")

# Input text for classification
text = "I love using Hugging Face models!"

# Classify text
result = classifier(text)
print("Classification Result:", result)
```

```python
from transformers import pipeline

# Load the NER pipeline
ner = pipeline("ner", grouped_entities=True)

# Input text for NER
text = "Hugging Face Inc. is a company based in New York City. Its headquarters are in DUMBO, therefore very close to the Manhattan Bridge."

# Perform NER
result = ner(text)
print("NER Result:", result)
```

```python
from transformers import pipeline

# Load the summarization pipeline
summarizer = pipeline("summarization")

# Input text for summarization
text = """Hugging Face is an AI company creating open-source tools for natural language processing. The company is known for its transformers library, which allows easy use of state-of-the-art models for tasks such as text classification, translation, and summarization."""

# Summarize text
summary = summarizer(text, max_length=50, min_length=25, do_sample=False)
print("Summary:", summary[0]['summary_text'])
```

```python
from transformers import pipeline

# Load the question answering pipeline
qa_pipeline = pipeline("question-answering")

# Define context and question
context = """Hugging Face Inc. is a company based in New York City. It is known for creating open-source tools for natural language processing."""
question = "Where is Hugging Face Inc. based?"

# Perform question answering
result = qa_pipeline(question=question, context=context)
print("Answer:", result['answer'])
```

```python
from transformers import pipeline

# Load the translation pipeline
translator = pipeline("translation_en_to_fr")

# Input text for translation
text = "Hugging Face is a company that focuses on natural language processing."

# Translate text to French
translation = translator(text, max_length=40)
print("Translation:", translation[0]['translation_text'])
```
