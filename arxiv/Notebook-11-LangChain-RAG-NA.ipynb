```python
from langchain.document_loaders import TextLoader
from langchain.indexers import FaissIndexer
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.chains import RAGChain
from langchain.llms import HuggingFaceLLM
```

```python
# Load documents
loader = TextLoader(directory_path="path_to_your_documents", file_extension="txt")
documents = loader.load()

# Create embeddings
embedder = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

# Create a vector store index
indexer = FaissIndexer(embedding=embedder)
indexer.add_documents(documents)
```

```python
# Initialize the language model
llm = HuggingFaceLLM(model_name="facebook/bart-large")
```

```python
# Initialize the RAG chain
rag_chain = RAGChain(
    retriever=indexer,
    llm=llm,
    question_key="question",  # Key for the input question
    documents_key="documents",  # Key for the retrieved documents
    output_key="response"  # Key for the generated response
)
```

```python
def generate_answer(question):
    # Use the RAG chain to generate an answer
    response = rag_chain.run({
        "question": question
    })
    return response["response"]
```

```python
question = "What are the benefits of using Retrieval-Augmented Generation?"
answer = generate_answer(question)
print(answer)
```
